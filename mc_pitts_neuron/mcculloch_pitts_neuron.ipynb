# McCulloch-Pitts Neuron

Build Boolean logic gates with a single threshold neuron. Run the cells from top to bottom.

```python
from itertools import product


def mp_neuron(x, types, theta):
    """Return the neuron's output for binary inputs and E/I/off connections."""
    if any(t == "I" and xi for xi, t in zip(x, types)):
        return 0
    count = sum(xi for xi, t in zip(x, types) if t == "E")
    return int(count >= theta)
```

## Logic gates

`E` is an excitatory connection, `I` is inhibitory, and `-` is disconnected. The third input is held at the indicated value for each gate.

```python
PAIRS = [(0, 0), (0, 1), (1, 0), (1, 1)]

# name: (connection types for x1,x2,x3, threshold, fixed x3, expected outputs)
GATES = {
    "AND            (E, E, theta=2)": ("EE-", 2, 0, [0, 0, 0, 1]),
    "OR             (E, E, theta=1)": ("EE-", 1, 0, [0, 1, 1, 1]),
    "x1 AND NOT x2  (E, I, theta=1)": ("EI-", 1, 0, [0, 0, 1, 0]),
    "NOR            (I, I, bias E=1, theta=1)": ("IIE", 1, 1, [1, 0, 0, 0]),
}


def table(types, theta, x3):
    return [mp_neuron([a, b, x3], types, theta) for a, b in PAIRS]
```

```python
print("x1 x2 ->", "  ".join(f"{a}{b}" for a, b in PAIRS), "\n")
for name, (types, theta, x3, expected) in GATES.items():
    out = table(types, theta, x3)
    assert out == expected, (name, out, expected)
    print(f"{name:<42} outputs: {out}")

print(
    "\nNOT x1 (x1=I, x3=bias E=1, theta=1):",
    [mp_neuron([a, 0, 1], "I-E", 1) for a in (0, 1)],
)
```

## XOR search

Check whether any combination of three excitatory, inhibitory, or disconnected inputs and thresholds from 0 to 3 produces XOR while `x3` is fixed at 0.

```python
xor = [a ^ b for a, b in PAIRS]
found = [
    (t, th)
    for t in product("EI-", repeat=3)
    for th in range(4)
    if table("".join(t), th, 0) == xor
]
print(f"XOR: {len(found)} of {27 * 4} settings work (x3 held at 0)")
```
